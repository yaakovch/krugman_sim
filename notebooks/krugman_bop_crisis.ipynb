# Krugman / Flood-Garber Balance-of-Payments Crisis

This notebook uses the same `krugman/` Python core as the Streamlit app. It shows the deterministic analytical solution, the numerical crossing check, core figures, and the Monte Carlo extension.

```python
from krugman.model import CrisisParams, attack_time, mechanical_exhaustion_time
from krugman.simulate import deterministic_simulation, monte_carlo_simulation
from krugman.plots import (
    core_figures,
    crisis_timing_curve_figure,
    mechanical_vs_attack_figure,
    monte_carlo_histogram_figure,
)

params = CrisisParams()
result = deterministic_simulation(params)
result.analytical_attack_time, result.numerical_attack_time, result.mechanical_exhaustion_time
```

## Closed-Form Timing

The fixed-rate reserve path is:

$$
R(t) = R_0 - \mu t
$$

Mechanical exhaustion occurs at:

$$
t_0 = \frac{R_0}{\mu}
$$

The post-collapse shadow floating exchange rate is:

$$
\tilde{s}(t) = D_0 + \mu t + \alpha(i^* + \mu)
$$

The rational attack occurs when the shadow float reaches the peg:

$$
\tilde{s}(t_c) = \bar{s}
$$

For an internally consistent peg, this reduces to:

$$
t_c = t_0 - \alpha
$$

```python
print(f"Mechanical exhaustion t0: {mechanical_exhaustion_time(params):.2f}")
print(f"Rational attack tc: {attack_time(params):.2f}")
print(f"Reserves just before attack: {result.reserves_before_attack:.2f}")
```

## Monte Carlo Extension

The stochastic version adds Gaussian shocks to domestic-credit growth. With zero shocks, it returns the deterministic crisis time exactly.

```python
mc = monte_carlo_simulation(params, n_paths=1000, sigma_credit=0.35, seed=1234)
mc.summary
```
